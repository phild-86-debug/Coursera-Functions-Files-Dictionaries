# Sorted The optional key parameter

```python
L1 = [1, 7, 4, -2, 3]

def absolute(x):
    if x >= 0:
        return x
    else:
        return -x

print(absolute(3))
print(absolute(-119))

for y in L1:
    print(absolute(y))
```

## Using the key parameter:  
> Now, we can pass the absolute function to sorted in order<br>
to specify that we want the items sorted in order of their absolute value,  <br>
rather than in order of their actual value.

```python
l1 = [17, 92, 12, 13, 47, 46, 101, 119, 181, 191]
# sort l1 in reverse order
l2 = sorted(l1, reverse=True)
print(l2)
l3 = [-100, 19, 5211, -2, -1001, 17, 182, -191, 1, -412]
# sort l3 in ascending order
l4 = sorted(l3)
print(l4)

# use absolute as the key
def absolute(x):
    if x < 0:
        return - x
    return x
print(absolute(-2))
print(l4)
print(sorted(l4,key=absolute))
l5 = sorted(l4, key=absolute)
print(l5)
```

#### Examples

```python
# sorted by second letter
ex_lst = ['hi', 'how are you', 'bye', 'apple', 'zebra', 'dance', 'g', 'q']
def second_let(string1):
    if len(string1) > 1:
        return string1[1]
    else:
        return ''
sorted_by_second_let = sorted(ex_lst, key=second_let)
print(sorted_by_second_let)
```

```python
# sorted by last letter then sorted by digit, highest to lowest
nums = ['1450', '33', '871', '19', '14378', '32', '1005', '44', '8907']

def last_char(nums):
    return nums[-1]

for num in nums:
    num = int(num)
nums_sorted = sorted(nums,key=last_char, reverse=True)
print(nums_sorted)
```

```python
# more difficult sorting - sorting by the last digit using a lambda expression
nums = ['1450', '33', '871', '19', '14378', '32', '1005', '44', '8907', '16']
def last_char(nums):
    return nums[-1]

for num in nums:
    num = int(num)

nums_sorted_lambda = (lambda x: sorted(x, key=last_char, reverse=True))(nums)
print(nums_sorted_lambda)
```
