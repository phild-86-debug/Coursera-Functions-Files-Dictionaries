```python
# using lambda- basic example
l1 = [1, 2, 18, 19, 3, 4, 101, 87, 81, 91, 11]
def absolute(x):
    if x >= 0:
        return x
    else:
        return -x


l2 = sorted(l1, key=lambda x: absolute(x))
print(l2)
```

```python
# Sorted by last char- uses function
nums = ['1450', '33', '871', '19', '14378', '32', '1005', '44', '8907', '16']

# def a function that returns the last char of a string
def last_char(num):
    ''' input: string .  output: last char of the string'''
    return num[-1]

 

# sort based on the last character of nums
y = sorted(nums, key=lambda x: last_char(x))
y
```

```python
#  sorting a dictionary based on key values using lambda expression

L = ['E', 'F', 'G', 'A', 'A', 'B', 'C', 'B', 'F', 'F', 'T', 'T', 'Q', 'C', 'Q', 'Q', 'Q', 'Q', 'L', 'F', 'Q', 'B']
l_dict = {}
for char in L:
    l_dict[char] = l_dict.get(char, 0) + 1

sorted_keys = sorted(l_dict, key=lambda x: l_dict[x], reverse=True)
print(l_dict)
print(sorted_keys)
for k in sorted_keys:
    print(k, ": ", l_dict[k])
```

```python
print(sorted(l_dict.keys()))
print(sorted(l_dict, key=lambda x: l_dict[x]))
```

```python
# sort dictionary using function
l2 = ['a', 'b', 'a', 'c', 'd', 'e', 'b', 'b', 'c', 'a', 'b', 'f', 'a', 'b', 'c', 'a', 'b', 'b', 'b', 'b', 'b']
l2_dict = {}
for char in l2:
    l2_dict[char] = l2_dict.get(char, 0) + 1


def g(k):
    return l2_dict[k]
sorted_dict = sorted(l2_dict.keys(), key=g, reverse=True)
print(sorted_dict)



```

```python
# Sorting Tie Breakers
fruits = ['peach', 'kiwi', 'apple', 'blueberry', 'papaya', 'mango', 'pear']
new_order = sorted(fruits, key=lambda fruit_name: (-len(fruit_name), fruit_name), reverse=True)
for fruit in new_order:
    print(fruit)

```

```python
# Sorted Weather
weather = {'Reykjavik': {'temp':60, 'condition': 'rainy'},
           'Buenos Aires': {'temp': 55, 'condition': 'cloudy'},
           'Cairo': {'temp': 96, 'condition': 'sunny'},
           'Berlin': {'temp': 89, 'condition': 'sunny'},
           'Caloocan': {'temp': 78, 'condition': 'sunny'}}

sorted_weather = sorted(weather, key=lambda w: (w, -weather[w]['temp']), reverse=True)
print(sorted_weather)
```
