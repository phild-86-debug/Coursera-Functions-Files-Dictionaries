# Sorting a dicitonary

```python
# Accumulate counts in a dictionary

L = ['E', 'F', 'B', 'A', 'D', 'I', 'I', 'C', 'B', 'A', 'D', 'D', 'E', 'D']
d = {}
for char in L:
    d[char] = d.get(char, 0) + 1
for x in d.keys():
    print("{} appears {} times.".format(x, d[x]))
```

```python
## Sorting Keys
L = ['E', 'F', 'B', 'A', 'D', 'I', 'I', 'C', 'B', 'A', 'D', 'D', 'E', 'D']

d = {}
for x in L:
    if x in d:
        d[x] = d[x] + 1
    else:
        d[x] = 1
y = sorted(d.keys())

for k in y:
    print("{} appears {} times".format(k, d[k]))

keys = d.keys()
print(keys)
print(sorted(keys))
```

### Sorted Dictionary based on count


```python
L = ['E', 'F', 'B', 'A', 'D', 'I', 'I', 'C', 'B', 'A', 'D', 'D', 'E', 'D']

d = {}
for x in L:
    if x in d:
        d[x] = d[x] + 1
    else:
        d[x] = 1

y = sorted(d.keys(), key=lambda k: d[k], reverse=True)
for k in y:
    print(k,":",  d[k])
```

```python
# Version of dictionary sorting that uses function as a key
L = ['E', 'F', 'B', 'A', 'D', 'I', 'I', 'C', 'B', 'A', 'D', 'D', 'E', 'D']

d = {}
for x in L:
    if x in d:
        d[x] = d[x] + 1
    else:
        d[x] = 1
def g(k):
    return d[k]

y = sorted(d.keys(), key=g, reverse=True)

for k in y:
    print(k, ": ", d[k])
```
