Write a function that takes a string instead of a list as a parameter and returns a list of the five most frequent characters in the string.

```python
def top_five(values):
    ''' Accepts a string.  Returns a  list of the five most frequent characters in the string '''
    l_dict = {}
    for val in values:
        l_dict[val] = l_dict.get(val, 0) + 1
    y = sorted(l_dict, key=lambda k: l_dict[k], reverse=True)
    return y[0:5]
    

top_five("dddeeeffffffqhijklmnoppppp")
```
